```python
import sys
from pathlib import Path
import pickle
import pandas as pd

# add repo root to python path
sys.path.append(str(Path().resolve().parent))

import core.models as Models
import core.engine_processor as Engine

with open("../data/records.pkl", "rb") as f:
    records = pickle.load(f)


rule_1 = {
    "name": "rule_1",
    "op": "AND",
    "conditions": [
        {"field": "firstinitial", "check": "match"},
        {"field": "firstname", "check": "not_nomatch"},
        {"field": "lastname", "check": "match"},
    ]
}

rule_2 = {
    "name": "rule_2",
    "op": "AND",
    "conditions": [
        {"field": "firstname", "check": "match"},
        {"field": "lastname", "check": "match"},
        {
            "op": "OR",
            "conditions": [
                {"field": "city", "check": "match"},
                {"field": "postalcode", "check": "match"},
            ],
        },
    ],
}

rules = [rule_1, rule_2]

def evaluate_rule(rule, datasource_result):
    if "field" in rule:
        field = rule["field"]
        check = rule["check"]

        state = datasource_result.fields.get(field, Models.MatchFieldState.MISSING)

        if check == "match":
            return state == Models.MatchFieldState.MATCH

        if check == "not_nomatch":
            return state != Models.MatchFieldState.NOMATCH

        raise ValueError(f"Unknown check type: {check}")

    op = rule["op"]
    conditions = rule["conditions"]

    if op == "AND":
        return all(evaluate_rule(child, datasource_result) for child in conditions)

    if op == "OR":
        return any(evaluate_rule(child, datasource_result) for child in conditions)

    raise ValueError(f"Unknown operator: {op}")


def find_valid_source_assignment(rule_index):

    if rule_index == len(rule_names):
        return True

    rule_name = rule_names[rule_index]
    ds_candidates = rule_results[rule_name]

    for ds in ds_candidates:
        if ds not in used_ds:
            rule_assignment[rule_name] = ds
            used_ds.add(ds)

            if find_valid_source_assignment(rule_index + 1):
                return True

            used_ds.remove(ds)
            del rule_assignment[rule_name]

    return False


outputs = []

for record in records[:500]:
    rule_results = {}

    for rule in rules:
        rule_name = rule["name"]
        passing_datasources = []

        for datasource_id, datasource_result in record.datasources.items():
            if evaluate_rule(rule, datasource_result):
                passing_datasources.append(datasource_id)

        rule_results[rule_name] = passing_datasources

    rule_names = list(rule_results.keys())
    used_ds= set()
    rule_assignment = {}

    verified = find_valid_source_assignment(0)

    outputs.append({
        "record_id": record.record_id,
        "verified": verified,
        "rule_results": rule_results,
        "rule_assignment": rule_assignment,
    })

summary_rows = []

for output in outputs:
    summary_rows.append({
        "record_id": output["record_id"],
        "verified": output["verified"],
        "rule_assignment": output["rule_assignment"],
        "rule_results": output["rule_results"],
    })

result_df = pd.DataFrame(summary_rows)

result_df
```
