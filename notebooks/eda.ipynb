```python
import sys
from pathlib import Path
import pickle
import pandas as pd
# add repo root to python path
sys.path.append(str(Path().resolve().parent))

import core.models as Models
import core.engine_processor as Engine

import core.engine_evaluator as Evaluator


# load sample records
with open("../data/records.pkl", "rb") as f:
    records = pickle.load(f)


rule_1 = {
    "name": "rule_1",
    "op": "AND",
    "conditions": [
        {"field": "firstinitial", "check": "match"},
        {"field": "firstname", "check": "not_nomatch"},
        {"field": "lastname", "check": "match"},
    ]
}


rule_2 = {
    "name": "rule_2",
    "op": "AND",
    "conditions": [
        {"field": "firstname", "check": "match"},
        {"field": "lastname", "check": "match"},
        {
            "op": "OR",
            "conditions": [
                {"field": "city", "check": "match"},
                {"field": "postalcode", "check": "match"},
            ],
        },
    ],
}


rules = [rule_1, rule_2]


# evaluate first 500 records
outputs = Evaluator.evaluate_records(
    records=records,
    rules=rules,
    limit=500
)


# convert to dataframe
summary_rows = []

for output in outputs:
    summary_rows.append({
        "record_id": output["record_id"],
        "verified": output["verified"],
        "rule_assignment": output["rule_assignment"],
        "rule_results": output["rule_results"],
    })


result_df = pd.DataFrame(summary_rows)


# simple score
verification_score = result_df["verified"].mean()

print(f"Verification Score: {verification_score:.2%}")


# preview results
result_df.head(10)
```

```python
def tokenize_rule(rule_text):

    rule_text = rule_text.lower().strip()

    tokens = re.findall(r"\(|\)|and|or|notnomatch|[a-zA-Z0-9_]+", rule_text)

    return tokens


def parse_rule(rule_text, rule_name):

    tokens = tokenize_rule(rule_text)

    rule_tree, next_index = parse_expression(tokens, 0)

    if next_index != len(tokens):
        raise ValueError(f"Unexpected token: {tokens[next_index]}")

    rule_tree = {
        "name": rule_name,
        **rule_tree
        }

    return rule_tree


# def parse_expression(tokens, index):

#     conditions = []

#     condition, index = parse_term(tokens, index)
#     conditions.append(condition)

#     while index < len(tokens) and tokens[index] in ["and", "or"]:
#         op = tokens[index].upper()
#         index += 1

#         next_condition, index = parse_term(tokens, index)

#         if len(conditions) == 1:
#             conditions = [
#                 {
#                     "op": op,
#                     "conditions": [
#                         conditions[0],
#                         next_condition
#                     ]
#                 }
#             ]

#         else:
#             current = conditions[0]

#             if current["op"] == op:
#                 current["conditions"].append(next_condition)
            
#             else:
#                 conditions = [
#                     {
#                         "op": op,
#                         "conditions": [
#                             current,
#                             next_condition
#                         ]
#                     }
#                 ]

#     return conditions[0], index


def parse_expression(tokens, index):

    current, index = parse_term(tokens, index)

    while index < len(tokens) and tokens[index] in ["and", "or"]:
        op = tokens[index].upper()
        index += 1

        next_condition, index = parse_term(tokens, index)

        if (
            isinstance(current, dict)
            and current.get("op") == op
        ):
            current["conditions"].append(next_condition)

        else:
            current = {
                "op": op,
                "conditions": [
                    current,
                    next_condition
                ]
            }

    return current, index




def parse_term(tokens, index):

    if tokens[index] == "(":
        index += 1

        expression, index = parse_expression(tokens, index)

        if index >= len(tokens) or tokens[index] != ")":
            raise ValueError("Missing closing parenthesis.")

        index += 1

        return expression, index

    if tokens[index] == "notnomatch":
        index += 1

        if index >= len(tokens):
            raise ValueError("Expected field after notnomatch.")

        field = tokens[index]
        index += 1

        return {
            "field": field,
            "check": "not_nomatch"
        }, index

    field = tokens[index]
    index += 1

    return {
        "field": field,
        "check": "match"
    }, index


def parse_rules(rule_texts):

    rules = []

    for i, rule_text in enumerate(rule_texts, start=1):
        rule = parse_rule(
            rule_text=rule_text,
            rule_name=f"rule_{i}"
        )

        rules.append(rule)

    return rules
```

```python
rule_texts = [
    """
    ( (firstinitial or firstname) and notnomatch lastname and 
    (dayofbirth and monthofbirth and yearofbirth) and 
    (address1 or (streetname and streetnumber and (city or postalcode))) )
    """,
    """
    ( firstinitial and notnomatch firstname and lastname and taxid )
    """
]


rules = parse_rules(rule_texts)
```
